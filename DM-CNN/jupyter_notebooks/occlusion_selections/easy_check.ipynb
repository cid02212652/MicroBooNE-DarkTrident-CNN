# Quick score browsing

Small helpers for finding interesting events in a score CSV or a folder of score CSVs.

```python
from pathlib import Path
from occlusion_selection_tools import (
    load_scores_any,
    select_top,
    nearest,
    filter_range,
    occlusion_cmd,
)

scores_path = Path(
    "/home/hep/an1522/dark_tridents_wspace/outputs/inference/run1_signal_resnet34_gn/run1_dt_ratio_0.6_ma_0.05_pi0_larcv_cropped_scores.csv"
)

df = load_scores_any(scores_path)
print("rows:", len(df))
display(df.head())
```

```python
select_top(df, n=10, by="signal_score")
```

```python
row = select_top(df, n=1, by="signal_score").iloc[0]
print("ENTRY:", int(row["entry_number"]))
print(
    "run/subrun/event:",
    int(row["run_number"]),
    int(row["subrun_number"]),
    int(row["event_number"]),
)
print("cmd:", occlusion_cmd(int(row["entry_number"])))
```

```python
nearest(df, col="signal_score", target=0.085, n=10)
```

```python
filter_range(df, score_min=0.05, pixels_min=1500).head(20)
```
